```python
import numpy as np

#making a matrix
myarr = np.arange(1,26)
mymatrix = myarr.reshape(5,5)
print(f"range: {myarr}")
print(f"mymatrix: \n{mymatrix}")

#multiplication
np.random.seed(1)
myarr2 = np.random.randint(1,21,size=(5,5))
print(f"random matrix: \n{myarr2}")
emult = np.multiply(mymatrix, myarr2)
print(f"element wise multiplication: \n{emult}")
dmult = mymatrix.dot(myarr2)
print(f"matrix wise multiplication: \n {dmult}")

#matrix addition
addmat = mymatrix + myarr2
print(f"matrix addition: \n{addmat}")

#summations 
sumcol = mymatrix.sum(axis=0)
sumrow = mymatrix.sum(axis=1)
print(f"printing summation col: \n{sumcol}")
print(f"printing summation row: \n{sumrow}")

#transposing 
tmatrix = mymatrix.T
print(f"printing transpose: \n{tmatrix}")

#splicing
print(f"printing middle 3x3: \n{mymatrix[1:4, 1:4]}")
print(f"printing row 2: \n{mymatrix[1]}")
print(f"prining col 3: \n{mymatrix[:,4]}")

```
